**idmax()** da el índice del valor máximo: `serie23.idxmax() # 4`

````python
# CÓDIGO DE LA SECCIÓN 2.3 SERIES: Accesos, conversiones y operaciones

print(f'\n{"="*15} 2.3 SERIES: Accesos, conversiones y operaciones')

serie23 = pd.Series(data=[0,1,2,3,4])
print(f'Serie original: {serie23}') # Imprime la serie original
print(f'\nHead(3): Imprime las primeras 3 filas de la serie\n{serie23.head(3)}') # Imprime las primeras 3 filas de la serie

print("\n2.3.1 Acceso a elementos de la serie usando diferentes métodos:")
print(f'Método at: {serie23.at[2]}')                     # 2.0
print(f'Método loc: {serie23.loc[2]}')                    # 2.0
print(f'Método iloc: {serie23.iloc[2]}')                   # 2.0
print(f'Método keys(): {serie23.keys()}')                # RangeIndex(start=0, stop=5, step=1

print("\n2.3.2 Conversiónes de tipo de datos")
serie23 = pd.Series.astype(serie23, dtype=np.float64)
print(f'A) Serie convertida a float64 con astype(): {serie23}') # Imprime la serie convertida a float64

print(f'\nB) Conversión de la serie a lista usando dos métodos diferentes: lo_list() y tolist()')

print(f'Conversión con to_list(): {pd.Series.to_list(serie23)}')
print(f'Conversión con tolist(): {serie23.tolist()}')                   # [0.0, 1.0, 2.0, 3.0, 4.0]

print("-"*40)

print(f'\n2.3.3 Reemplazar where() y mask()')

print("Ahora solo se imprime el resultado de where y mask; no se modifica la serie original")
# Reemplazar donde la condición NO se cumple
print("\nA) Método where(): reemplaza donde la condición NO se cumple")
print("Resultado de la serie original después de usar where: reemplaza los valores mayores a 2 por 0")
print(serie23.where(serie23 < 2, other=0)) # Reemplaza los valores menores a 2 por 0

# Reemplazar donde la condición SÍ se cumple
print("\nB) Método mask(): reemplaza donde la condición SÍ se cumple")
print("Resultado de la serie original después de usar mask: reemplaza los valores menores a 2 por 5")
print(serie23.mask(serie23 < 2, other=5))  # Reemplaza los valores menores a 2 por 5

print("\n2.3.4 Operaciones con Series")

print("A) Operaciones aritméticas entre series:")
serie234a = pd.Series([0,1,2,3,4])
serie234b = pd.Series([5,6,7,8,9])

print(f'Suma de dos series:\n{serie234a + serie234b}')         # Suma elemento a elemento
print(f'Resta de dos series:\n{serie234a - serie234b}')      # Resta elemento a elemento
print(f'Multiplicación de dos series:\n{serie234a * serie234b}') # Multiplicación elemento a elemento
print(f'División de dos series:\n{serie234a / serie234b}')      # División elemento a elemento

print("\nB) Concatenación de series:")
serie234c = pd.concat([serie234a, serie234b])
print(f'Concatenación de las dos series:\n{serie234c}')

print("\nC) Métodos estadísticos")
print(f'idmax imprime el índice del valor máximo: {serie23.idxmax()}') # 4
print(f'idmin imprime el índice del valor mínimo: {serie23.idxmin()}') # 0
print(f'max imprime el valor máximo: {serie23.max()}') # 4.0
print(f'min imprime el valor mínimo: {serie23.min()}')  # 0.0
print(f'sum imprime la suma de los valores: {serie23.sum()}')  # 10.0
print(f'mean imprime el valor promedio: {serie23.mean()}')  # 2.0

````

### 2.4.3 Método isna()

El método `pd.Series.isna()` sirve para detectar valores **nulos o NaN** en una serie.  
Uso típico: Es fundamental para la limpieza de datos antes del análisis. 

Devuelve una nueva Serie de tipo booleano (True/False) con la misma forma.   
    *True: Si el valor es NaN (Not a Number), None o NaT (Not a Time).
    *False: Si el valor es válido (datos numéricos, cadenas, etc.).
    
**Diferencia con isnull()**: isna() e isnull() funcionan exactamente de la misma manera; isnull() es simplemente un alias de isna().

### 2.5.2 Serie creada con un diccionario de Series a DataFrame

Si la serie tiene **índices diferentes** (por ejemplo, 1, 2) y el DataFrame tiene otros (0, 1), **pandas creará filas adicionales** para acomodar todos los índices.
Si la serie es más corta o larga, **los valores faltantes se llenarán con NaN**.

DataFrame original

`df_diccionario = pd.DataFrame({'A': [1, 2]}, index=[0, 1])`

Serie creada desde diccionario (índices 1 y 2):

